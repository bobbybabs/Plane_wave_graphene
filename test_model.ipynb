```python
a=np.linalg.norm(model.cell[:,1])
n=model.icell[0]/np.linalg.norm(model.icell[0])
def landau_gauge(r,p):
    return 2*np.pi*r[1]*p/model.q

A=landau_gauge(model.r,p=3)
phase=np.exp(1j*A*model.q)
x,y=np.einsum('id,ixy->dxy',model.cell.T,model.r,optimize=True)
plt.figure(figsize=(8,2))
plt.scatter(x,y,s=4,c=phase.imag)
plt.axis('equal')
plt.colorbar()
plt.show()
```

```python
g=np.linalg.norm(model.G,axis=-1)
plt.figure()
plt.scatter(g,model.fft(A).imag)
plt.scatter(g,model.fft(A).real)
plt.show()
```

```python
model.plot_mesh(c=model.fft(A).real)
```
